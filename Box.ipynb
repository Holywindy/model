```python
import seaborn as sns
import numpy as np
import pandas as pd
from pandas import Series,DataFrame
import matplotlib.pyplot as plt

file_path='E:/google/data5.csv'
s=pd.read_csv(file_path)
num=24

data = None
for i in range(31):
    cpu=s.iloc[:,i*2+1].values.tolist()
    memory=s.iloc[:,i*2+2].values.tolist()
    cpu_type = ['cpu' for _ in range(num)] 
    memory_type = ['memory' for _ in range(num)] 
    day = [i+1 for _ in range(num)] 
    cpu = pd.DataFrame({'Percentage': cpu, 'type': cpu_type, 'day': day}) 
    memory = pd.DataFrame({'Percentage': memory, 'type': memory_type, 'day': day}) 
    if i == 0:
        data = pd.concat([cpu,memory])
    else:        
        data = pd.concat([data,cpu,memory])

plt.figure(figsize=(30,6))
sns.boxplot(x="day", y="Percentage", data=data, hue="type", width=0.5, linewidth=1.0, palette="Set3")

plt.title("Comparison of Cpu usage and Memory usage over 31days",fontsize=18)
plt.xlabel("Days",fontsize=18)
plt.ylabel("Percentage",fontsize=18)
num1 = 1.001
num2 = 0.9
num3 = 3
num4 = 0
plt.legend(bbox_to_anchor=(num1, num2), loc=num3, borderaxespad=num4)
plt.tick_params(labelsize=18)
#lt.ylim(0, 100)
plt.show()
```

```python
import pandas as pd
import pandas_datareader
import datetime
import matplotlib.pylab as plt
import seaborn as sns
from matplotlib.pylab import style
from statsmodels.tsa.arima_model import ARIMA
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

style.use('ggplot')    
plt.rcParams['font.sans-serif'] = ['SimHei'] 
plt.rcParams['axes.unicode_minus'] = False  
stockFile = 'E:/google/b3_stocks_1994_2020.csv' 
stock = pd.read_csv(stockFile, index_col=0, parse_dates=[0])

stock_week = stock['close'].resample('W-MON').mean()
stock_train = stock_week['2013':'2020']
stock_train.plot(figsize=(12,8))
plt.legend(bbox_to_anchor=(1, 1))
plt.title("Stock Close")
sns.despine()

stock_diff = stock_train.diff()
stock_diff = stock_diff.dropna()

plt.figure()
stock_diff.plot(figsize=(12,8))
plt.title('First-order differential')
plt.show()

acf = plot_acf(stock_diff, lags=20)
plt.title("ACF")
acf.show()

pacf = plot_pacf(stock_diff, lags=20)
plt.title("PACF")
pacf.show()
```
