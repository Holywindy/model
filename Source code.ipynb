```python
import numpy as np
import matplotlib
import matplotlib as mpl
import matplotlib.pyplot as plt

def heatmap(data, row_labels, col_labels, ax=None,
            cbar_kw={}, cbarlabel="", **kwargs):


    if not ax:
        ax = plt.gca()
   

    
    im = ax.imshow(data, **kwargs)
    
    cbar = ax.figure.colorbar(im, ax=ax, **cbar_kw)
    
    cbar.ax.set_ylabel(cbarlabel, rotation=-90, va="bottom")
    
    ax.set_xticks(np.arange(data.shape[1]))
    ax.set_yticks(np.arange(data.shape[0]))
   
    ax.set_xticklabels(col_labels)
    ax.set_yticklabels(row_labels)
   
    for edge, spine in ax.spines.items():
        spine.set_visible(False)

    ax.set_xticks(np.arange(data.shape[1]+1)-.5, minor=True)
    ax.set_yticks(np.arange(data.shape[0]+1)-.5, minor=True)
   
    ax.grid(which="minor", color="w", linestyle='-', linewidth=3)
   
    ax.tick_params(which="minor", bottom=False, left=False)
 
    

    return im, cbar


def annotate_heatmap(im, data=None, valfmt="{x:.2f}",
                     textcolors=["black", "white"],
                     threshold=None, **textkw):


    if not isinstance(data, (list, np.ndarray)):
        data = im.get_array()
   
    if threshold is not None:
        threshold = im.norm(threshold)
    else:
        threshold = im.norm(data.max())/2.

   
    kw = dict(horizontalalignment="center",
              verticalalignment="center")
    kw.update(textkw)

   
    if isinstance(valfmt, str):
        valfmt = matplotlib.ticker.StrMethodFormatter(valfmt)
   
    texts = []
    for i in range(data.shape[0]):
        for j in range(data.shape[1]):
            kw.update(color=textcolors[int(im.norm(data[i, j]) > threshold)])
            text = im.axes.text(j, i, valfmt(data[i, j], None), **kw)
            texts.append(text)
```

```python
import numpy as np

def calculate_acf(time_series, max_lag=100):
    """
    Calculate the Autocorrelation Function (ACF) for a given time series and lags up to max_lag.
    
    Parameters:
    - time_series: numpy array of the time series data
    - max_lag: the maximum lag to which the ACF is computed
    
    Returns:
    - acf_values: numpy array containing the ACF values from lag 1 to max_lag
    """
    n = len(time_series)
    data_mean = np.mean(time_series)
    c0 = np.sum((time_series - data_mean) ** 2) / n

    acf_values = np.array([1.0] + [np.corrcoef(time_series[:-lag], time_series[lag:])[0, 1] for lag in range(1, max_lag)])
    return acf_values

def mse_acf(true_series, generated_series, max_lag=100):
    """
    Compute the Mean Squared Error (MSE) of ACF values between the true time series and the generated time series.
    
    Parameters:
    - true_series: numpy array of the true time series data
    - generated_series: numpy array of the generated time series data
    - max_lag: the maximum lag to which the ACF is computed
    
    Returns:
    - mse_value: the MSE of ACF values
    """
    true_acf = calculate_acf(true_series, max_lag)
    gen_acf = calculate_acf(generated_series, max_lag)
    mse_value = np.mean((true_acf - gen_acf) ** 2)
    return mse_value

# Example usage:
# true_series = np.array(...) # True time series data
# generated_series = np.array(...) # Generated time series data

# Replace the above lines with actual data to calculate the MSE of ACF values

# mse_value = mse_acf(true_series, generated_series)
# print(f"The MSE of ACF values is: {mse_value}")
```

```python
import numpy as np

def manual_acf(data, max_lag):
    n = len(data)
    data_mean = np.mean(data)
    var = np.var(data)
    
    acfs = []
    for k in range(1, max_lag + 1):
        cov = np.sum((data[:-k] - data_mean) * (data[k:] - data_mean)) / (n - k)
        acf = cov / var
        acfs.append(acf)
    
    return acfs

# 创建一个示例时间序列数据
np.random.seed(0)
data = np.random.randn(100).cumsum()

# 计算自相关函数
max_lag = 100
acf_values = manual_acf(data, max_lag)

# 输出自相关系数
print(acf_values)
print(data)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 数据
labels = ['ACF', 'Skew', 'Kurt']
models = ['WGAN-gp', 'TimeGAN', 'GT-GAN', 'MT-GAN', 'AEC-GAN', 'Ours']
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']  # 每个模型对应一个颜色

data_2019_google = [
    [0.579, 8.66, 75.54],  # WGAN-gp
    [0.715, 13.52, 177],   # TimeGAN
    [0.443, 3.588, 27.24], # GT-GAN
    [0.84, 1.69, 3.85],    # MT-GAN
    [0.877, 3.876, 18.29], # AEC-GAN
    [0.164, 1.365, 1.57]   # ours
]

# 雷达图的角度数
angles = np.linspace(0, 2 * np.pi, len(labels), endpoint=False).tolist()
angles += angles[:1]  # 闭合雷达图

# 标准化数据（按最大值进行归一化）
data_2019_google_normalized = np.array(data_2019_google)
data_2019_google_normalized = data_2019_google_normalized / data_2019_google_normalized.max(axis=0)

# 画图函数
def plot_radar(data, title, ax, colors):
    for idx, model_data in enumerate(data):
        model_data = np.concatenate((model_data, model_data[:1]))  # 闭合图形
        model_data = 1 - model_data  # 值越小越靠近中心
        ax.fill(angles, model_data, alpha=0.25, color=colors[idx])
        ax.plot(angles, model_data, linewidth=2, color=colors[idx])

    ax.set_yticklabels([])  # 去掉 y 轴标签
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(labels,fontsize=18)
    ax.set_title(title, size=24, color='black')

# 创建图形
fig, ax = plt.subplots(figsize=(6, 6), dpi=100, subplot_kw=dict(polar=True))

# 绘图
plot_radar(data_2019_google_normalized, '2019 Google Performance', ax, colors)

# 设置 y 轴
ax.set_ylim(0, 1)
ax.set_yticks([0, 0.2, 0.4, 0.6, 0.8, 1])

# 添加图例并对应颜色
for idx, label in enumerate(models):
    ax.plot([], [], color=colors[idx], label=label)  # 画一个隐藏的图例用的标识
ax.legend(loc='upper right', bbox_to_anchor=(1.4, 1), prop={'size': 14})

plt.show()
```
